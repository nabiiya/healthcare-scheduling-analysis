```python
#converts values into dates and saves it back into the same column
appointments["scheduling_date"] = pd.to_datetime(appointments["scheduling_date"])
appointments["appointment_date"] = pd.to_datetime(appointments["appointment_date"])
```

```python
#creates new column called appointment_day and saves weekday names into it
appointments["appointment_day"] = appointments["appointment_date"].dt.day_name()
```

```python
appointments[["appointment_date", "appointment_day"]].head()
```

```python
#extracting the hour
appointments["appointment_hour"] = pd.to_datetime(
    appointments["appointment_time"],
    format="%H:%M:%S"
).dt.hour
```

```python
appointments[["appointment_time", "appointment_hour"]].head()
```

```python
#categorize appointment times to create morning, afternoon, and evening
def get_time_period(hour):
    if hour < 12:
        return "Morning"
    elif hour < 17:
        return "Afternoon"
    else:
        return "Evening"
```

```python
#applying the function to the whole appoint_hour column
appointments["time_period"] = appointments["appointment_hour"].apply(get_time_period)
```
